<h2 style="color: #1D4ED8; font-size: 20px; font-weight: bold; margin: 5px 0 10px;">
    📈 Scatter Plot: Study Hours and Final Exam Grades
</h2>

<p style="font-size: 16px; margin: 6px 0;">
    <b>Scenario:</b> A statistics instructor wants to examine
    whether a relationship exists between the number of hours
    students spend studying for their final exam (Hours)
    and their final exam grades (Grade).
    She collects data from a sample of eight students
    to investigate the relationship between these two variables.
</p>

</div>



<div style="font-family: 'Times New Roman', serif;">

<h3 style="color: #1D4ED8; font-size: 19px; font-weight: bold; margin: 12px 0 5px;">
    Task 01: Data Preparation Using Pandas
</h3>

<ul style="font-size: 16px; line-height: 1.5; padding-left: 20px;">
    <li>
        Create a Pandas DataFrame using the study hours (Hours)
        and final exam grades (Grade) of eight students.
        Display the first five rows.
    </li>
</ul>

</div>


```python
# Step 1: Import Pandas
import pandas as pd

# Step 2: Create a dataset
data = {
    'Hours': [8, 2, 3, 8, 10, 15, 25, 5],
    'Grade': [75, 47, 50, 80, 85, 88, 93, 55]
}

# Step 3: Create a DataFrame
df = pd.DataFrame(data)

# Step 4: Display the first 5 rows
print(df.head())
```

<div style="font-family: 'Times New Roman', serif;">

<h3 style="color: #1D4ED8; font-size: 19px; font-weight: bold; margin: 12px 0 5px;">
    Task 02: Scatter Plot Using Matplotlib
</h3>

<ul style="font-size: 16px; line-height: 1.5; padding-left: 20px;">
    <li>
        Create a scatter plot using Matplotlib, with study hours
        (Hours) on the X-axis and final exam grades (Grade)
        on the Y-axis. Add an appropriate title and axis labels.
    </li>
</ul>

</div>

```python
# Step 1: Import libraries
import matplotlib.pyplot as plt
import numpy as np

# Step 2: Create a scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(df['Hours'], df['Grade'], color='blue')

# Step 3: Add a trendline
m, b = np.polyfit(df['Hours'], df['Grade'], 1)
plt.plot(df['Hours'], m * df['Hours'] + b, color='red')

# Step 4: Add title and axis labels
plt.title('Study Hours vs Final Exam Grades')
plt.xlabel('Study Hours')
plt.ylabel('Final Exam Grades')

# Step 5: Display the scatter plot
plt.show()
```


<div style="font-family: 'Times New Roman', serif; font-size: 12px;">

<h4 style="font-size: 16px; font-weight: bold; margin: 5px 0;">
    Understanding Python Syntax
</h4>

<ul style="font-size: 12px; line-height: 1.5; padding-left: 20px;">
    <li><code>import matplotlib.pyplot as plt</code> – Imports Matplotlib.</li>
    <li><code>import numpy as np</code> – Imports NumPy.</li>
    <li><code>plt.figure(figsize=(8, 5))</code> – Sets the chart size.</li>
    <li><code>plt.scatter()</code> – Creates a scatter plot.</li>
    <li><code>df['Hours']</code> – Selects study hours for the X-axis.</li>
    <li><code>df['Grade']</code> – Selects exam grades for the Y-axis.</li>
    <li><code>color='blue'</code> – Sets the scatter plot points to blue.</li>
    <li><code>np.polyfit()</code> – Calculates the slope and intercept of the best-fit line.</li>
    <li><code>m, b</code> – Stores the slope (m) and intercept (b).</li>
    <li><code>plt.plot()</code> – Draws the trendline.</li>
    <li><code>color='red'</code> – Sets the trendline color to red.</li>
    <li><code>plt.title()</code> – Adds the chart title.</li>
    <li><code>plt.xlabel()</code> – Labels the X-axis.</li>
    <li><code>plt.ylabel()</code> – Labels the Y-axis.</li>
    <li><code>plt.show()</code> – Displays the scatter plot.</li>
</ul>

</div>



<div style="font-family: 'Times New Roman', serif; text-align: center;">

<h1 style="color: #0F766E; font-size: 32px; font-weight: bold; margin: 8px 0;">
    📊 Scatter Plot Using Python: Seaborn
</h1>

</div>

```python
# Step 1: Import libraries
import seaborn as sns
import matplotlib.pyplot as plt

# Step 2: Create a scatter plot with a trendline
plt.figure(figsize=(8, 5))
sns.regplot(data=df, x='Hours', y='Grade',
            scatter_kws={'color': 'blue'},
            line_kws={'color': 'red'},
            ci=None)

# Step 3: Add title and axis labels
plt.title('Study Hours vs Final Exam Grades')
plt.xlabel('Study Hours')
plt.ylabel('Final Exam Grades')

# Step 4: Display the scatter plot
plt.show()
```

    Understanding Python Syntax

    <li><code>import seaborn as sns</code> – Imports Seaborn.</li>
    <li><code>import matplotlib.pyplot as plt</code> – Imports Matplotlib.</li>

    <li><code>scatter_kws={'color': 'blue'}</code> – Sets the scatter plot points to blue.</li>

    <li><code>ci=None</code> – Removes the shaded confidence interval.</li>

<h1 style="color: #7E22CE; font-size: 32px; font-weight: bold; margin: 8px 0;">
    📊 Scatter Plot Using Python: Plotly
</h1>

</div>


```python

# Step 1: Import Plotly
import plotly.express as px

# Step 2: Create a scatter plot with a trendline
fig = px.scatter(
    df,
    x='Hours',
    y='Grade',
    trendline='ols',
    color_discrete_sequence=['blue'],
    title='Study Hours vs Final Exam Grades',
    width=800,
    height=500
)

# Step 3: Add axis labels
fig.update_layout(
    xaxis_title='Study Hours',
    yaxis_title='Final Exam Grades'
)

# Step 4: Display the scatter plot
fig.show()
```

    <b>Pass Completion Rate (PCR %)</b>,

</h3>

<p style="font-size: 16px; line-height: 1.5;">
    Import <b>Football_Game.xlsx</b> using Pandas, display the first five rows, check for missing values, and review the dataset information.
</p>

</div>

```python
# Step 1: Import Pandas
import pandas as pd

# Step 2: Read the Excel file
df = pd.read_excel('Football_Game.xlsx')

# Step 3: Display the first 5 rows
print(df.head())

# Step 4: Check missing values
print(df.isnull().sum())

# Step 5: Check dataset information
df.info()
```


<div style="font-family: 'Times New Roman', serif;">

<h3 style="color: #1D4ED8; font-size: 19px; font-weight: bold; margin: 12px 0 5px;">
    Task 02: Scatter Plot Using Matplotlib

<p style="font-size: 16px; line-height: 1.5;">
    Create a scatter plot using Matplotlib to examine the relationship between <b>Pass Completion Rate (PCR %)</b> on the X-axis and <b>Salary ($ millions)</b> on the Y-axis. Add an appropriate title, axis labels, and a linear trendline.
</p>

</div>


```python
# Step 1: Import libraries
import matplotlib.pyplot as plt
import numpy as np

# Step 2: Create a scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(df['PCR'], df['Salary'], color='blue')

# Step 3: Add a linear trendline
m, b = np.polyfit(df['PCR'], df['Salary'], 1)
plt.plot(df['PCR'], m * df['PCR'] + b, color='red')

# Step 4: Add title and axis labels
plt.title('Pass Completion Rate vs Quarterback Salary')
plt.xlabel('Pass Completion Rate (%)')
plt.ylabel('Salary ($ Millions)')

# Step 5: Display the scatter plot
plt.show()
```

    <li><code>import matplotlib.pyplot as plt</code> – Imports Matplotlib for creating plots.</li>
    <li><code>import numpy as np</code> – Imports NumPy for numerical calculations.</li>
    <li><code>plt.figure(figsize=(8, 5))</code> – Sets the chart size to 8 × 5 inches.</li>
    <li><code>plt.scatter()</code> – Creates a scatter plot.</li>

    <li><code>color='blue'</code> – Sets the scatter plot points to blue.</li>
    <li><code>np.polyfit(df['PCR'], df['Salary'], 1)</code> – Calculates the slope and intercept of a linear trendline.</li>
    <li><code>m, b</code> – Stores the slope (m) and intercept (b).</li>
    <li><code>plt.plot()</code> – Draws the red linear trendline.</li>
