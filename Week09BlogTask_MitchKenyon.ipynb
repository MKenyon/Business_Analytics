```python
#import the necessary libraries
import sqlite3
import pandas as pd

#connect to the SQLite database (or create it if it doesn't exist)
conn = sqlite3.connect('my_data.db')

#create a cursor object to interact with the database
cursor = conn.cursor()
```

1. SELECT: Sales Data
Display sale_date, product, and price from the sales table.

```python
#read the sales.csv file and write it to the sales table in the database
pd.read_csv('sales.csv').to_sql('sales', conn, if_exists='replace', index=False)

#query the sales table to retrieve the sale_date, product, and price columns
pd.read_sql_query('''
SELECT sale_date, product, price FROM sales
''', conn)
```

2. WHERE: Human Resources
Find employees in Chicago with more than 10 years of experience using WHERE and AND.

```python
#read the hr_employees.csv file and write it to the hr_employees table in the database
pd.read_csv('hr_employees.csv').to_sql('hr_employees', conn, if_exists='replace', index=False)

#query the hr_employees table to retrieve all columns for employees in Chicago with more than 10 years of experience
pd.read_sql_query('''
SELECT *
FROM employees
WHERE city = 'Chicago'
  AND years_experience > 10;
''', conn)
```

3. ORDER BY: Real Estate 
Display the three largest properties by square footage using ORDER BY and LIMIT.

```python
#read the realestate_listings.csv file and write it to the listings table in the database
pd.read_csv('realestate_listings.csv').to_sql('listings', conn, if_exists='replace', index=False)

#query the listings table to retrieve the neighborhood, property_type, list_price, and sqft columns for the three largest properties by square footage
pd.read_sql_query('''
SELECT neighborhood, property_type, list_price, sqft
FROM listings
ORDER BY sqft DESC LIMIT 3
''', conn)
```

4. GROUP BY: Marketing
Calculate total weekly impressions using SUM() and GROUP BY.

```python
#read the marketing_campaigns.csv file and write it to the campaigns table in the database
pd.read_csv('marketing_campaigns.csv').to_sql('campaigns', conn, if_exists='replace', index=False)

#query the campaigns table to retrieve the campaign name and total spend for each campaign, grouped by campaign
pd.read_sql_query('''
SELECT campaign, SUM(impressions) AS total_spend
FROM campaigns
GROUP BY campaign
''', conn)
```

5. HAVING: Restaurant
Find dates when total restaurant sales exceeded $400.

```python
#read the clinic_visits.csv file and write it to the visits table in the database
pd.read_csv('restaurant_orders.csv').to_sql('orders', conn, if_exists='replace', index=False)

#query the orders table to retrieve the order_date and total_sales for each day, grouped by order_date, where total_sales is greater than 400
pd.read_sql_query('''
SELECT order_date, ROUND(SUM(order_total), 2) 
AS total_sales
FROM orders
GROUP BY order_date
HAVING SUM(order_total) > 400
''', conn)
```

6. Aggregate Functions: Healthcare
Find the average patient age and longest waiting time for each doctor.

```python
#read the clinic_visits.csv file and write it to the visits table in the database
pd.read_csv('clinic_visits.csv').to_sql('visits', conn, if_exists='replace', index=False)

#query the visits table to retrieve the doctor, average patient age, and maximum wait time for each doctor, grouped by doctor
pd.read_sql_query('''
SELECT doctor,
AVG(ROUND(patient_age,2)) AS avg_patient_age,
MAX(wait_minutes) AS max_wait_minutes
FROM visits
GROUP BY doctor
''', conn)
```

7. JOIN: Banking
Join customer and transaction tables to calculate total transaction amounts by branch.

```python
#read the bank customers and transactions CSV files into the database
pd.read_csv('bank_customers.csv').to_sql('customers', conn, if_exists='replace', index=False)
pd.read_csv('bank_transactions.csv').to_sql('transactions', conn, if_exists='replace', index=False)

#query to calculate the total transaction amount for each branch
pd.read_sql_query('''
SELECT c.branch,
SUM(t.amount) AS total_transaction_amount
FROM transactions AS t
JOIN customers AS c ON t.customer_id = c.customer_id
GROUP BY c.branch
''', conn)
```
