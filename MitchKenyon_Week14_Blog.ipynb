```python
# Import Libraries
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# Load File
df = pd.read_excel('applications.xlsx') 

# Display first 5 rows
print(df.head)

# Step 4: Check missing values
print(df.isnull().sum())
```

Task 01: Bar Plot: Compare the number of applications received by each college. Which college attracts the most applicants?

The college that attracts the most applicants is the College of Arts and Letters. 

Matplotlib Bar Plot

```python
# Step 1: Count Values
app_counts = df['College'].value_counts()

# Step 2: Create Bar Graph
plt.bar(app_counts.index, app_counts.values)

# Step 3: Create Title for the Graph
plt.title('College Applications')

# Step 4: Label x axis
plt.xlabel('College')

# Step 5: Label y axis
plt.ylabel('Number of Applications')

# Step 6: Print Graph
plt.show()
```

Seaborn Bar Plot

```python

# Step 1: Count applicants by college
app_counts = df['College'].value_counts()

# Step 2: Create Bar Graph
sns.barplot(x=app_counts.index, y=app_counts.values, color='darkgrey')

# Step 3: Create Title for the Graph
plt.title("Applications by College")

# Step 4: Label x axis
plt.xlabel('College')

# Step 5: Label y axis
plt.ylabel("Number of Applications")

# Step 6: Print Graph
plt.show()
```

Plotly Bar Plot

```python
# Step 1: Count applicants by college
app_counts = df['College'].value_counts()

# Step 2: Create Bar Graph
fig = px.bar(x=app_counts.index, y=app_counts.values,
             color_discrete_sequence=['darkgrey'],
             title="Applications by College")


# Step 3: Create Title for the Graph
fig.update_layout(
    xaxis_title = 'College',
    yaxis_title='Number of Applications'
)

# Step 4: Print Graph
fig.show()

```

Task 02: Histogram: Visualize the distribution of SAT_ACT scores. What is the most common score range?

The most common score range is 1100 to 1200

Matplotlib Histogram

```python
# Step 1: Set figure size
plt.figure(figsize=(8, 5))

# Step 2: Create Histogram
plt.hist(df['SAT_ACT'], bins=10,
         color='darkgrey')

# Step 3: Create Title for the Graph
plt.title("Distribution of SAT_ACT Scores")

# Step 4: Label x axis
plt.xlabel("SAT_ACT Scores")

# Step 5: Label y axis
plt.ylabel("Number of Applicants")

# Step 6: Print Graph
plt.show()
```

Seaborn Histogram

```python
# Step 1: Set figure size
plt.figure(figsize=(8, 5))

# Step 2: Create Histogram
sns.histplot(df['SAT_ACT'], bins=10,
         color='red')

# Step 3: Create Title for the Graph
plt.title("Distribution of SAT_ACT Scores")

# Step 4: Label x axis
plt.xlabel("SAT_ACT Scores")

# Step 5: Label y axis
plt.ylabel("Number of Applicants")

# Step 6: Print Graph
plt.show()
```

Plotly Histogram

```python
# Step 1: Create Histogram
fig = px.histogram(
    df,
    x='SAT_ACT',
    nbins=10,
    color_discrete_sequence=['darkgrey'],
    title='Distribution of SAT_ACT Scores',
    width=800,
    height=500
)

# Step 2: Create axis for the Graph
fig.update_layout(
    xaxis_title='SAT_ACT Scores',
    yaxis_title='Number of Applicants'
)

# Step 3: Print Graph
fig.show()
```

Task 03: Scatter Plot: Visualize the relationship between SAT_ACT scores and Extracurricular_Hours. Is there a positive, negative, or no clear relationship?

Since there is no upward or downward trend on the scatterplot, there is no clear relationship between SAT_ACT scores and extracurricular activities. 

Matplotlib Scatter Plot

```python
# Step 1: Create Chart Size
plt.figure(figsize=(8,5))

# Step 2: Create Scatter Plot
plt.scatter(df["Extracurricular_Hours"], df["SAT_ACT"], color="red")

# Step 3: Create a trendline
m,b= np.polyfit(df["Extracurricular_Hours"],df["SAT_ACT"], 1)
plt.plot(df["Extracurricular_Hours"], m*df["Extracurricular_Hours"]+b, color= "black")

# Step 4: Create a title
plt.title("SAT and ACT Scores vs. Extracurricular Hours")

# Step 4: Name the Axis
plt.xlabel("Extracurricular Hours per Week")
plt.ylabel("SAT and ACT Score")

# Step 6: Print the scatter plot
plt.show()
```

Seaborn Scatter Plot

```python
# Step 1: Create Chart Size
plt.figure(figsize=(8,5))

# Step 2: Create Scatter Plot
sns.regplot(data=df, x="Extracurricular_Hours", y="SAT_ACT",
            scatter_kws={"color":"red"},
            line_kws={"color":"black"},
            ci=None)

# Step 4: Create a title
plt.title("SAT and ACT Scores vs. Extracurricular Hours")

# Step 4: Name the Axis
plt.xlabel("Extracurricular Hours per Week")
plt.ylabel("SAT and ACT Score")

# Step 5: Print the scatter plot
plt.show()
```

Plotly Scatter Plot

```python
# Step 1: Create a scatter plot with a trendline
fig = px.scatter(
    df,
    x="Extracurricular_Hours",
    y="SAT_ACT",
    trendline="ols",
    color_discrete_sequence=["red"],
    title="SAT and ACT Scores vs. Extracurricular Hours",
    width=800,
    height=500
)

# Step 2: Add axis labels
fig.update_layout(
    xaxis_title='Extracurricular Hours per Week',
    yaxis_title='SAT and ACT Score')

# Step 3: Display the scatter plot
fig.show()
```

1. Arts & Letters Receives the Most Applications

Recommendation: Allocate recruitment resources according to application demand. Continue supporting Arts & Letters while increasing targeted marketing and outreach for Business & Economics to attract more applicants.

2. Most Applicants Have SAT_ACT Scores Between 1,000 and 1,399
The average SAT_ACT score is approximately 1,165, with the largest concentration of applicants scoring between 1,100 and 1,199.

Recommendation: Use SAT_ACT score distributions to improve academic advising, scholarship planning, and recruitment strategies. Consider additional academic support for students with lower scores while offering competitive scholarships to attract high-performing applicants.

3. Enrollment Is Significantly Lower Than Applications
Of the 17,339 applicants, 5,323 were admitted and 1,528 were recorded as enrolled. Math & Science received 6,272 applications but recorded only 488 enrollments.

Recommendation: Improve enrollment conversion by strengthening communication with admitted students, offering financial aid guidance, and providing personalized follow-ups. Focus particularly on Math & Science to identify barriers preventing applicants from enrolling.

Overall Recommendation
The institution should focus on targeted recruitment, data-driven academic support, and improved enrollment conversion. By analyzing application volume, SAT_ACT performance, and enrollment outcomes, admissions teams can better allocate resources, improve student engagement, and develop more accurate enrollment forecasts.
